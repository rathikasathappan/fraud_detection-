# Real-Time Cold-Start Fraud Detection: Exploratory Data Analysis
This notebook analyzes the merchant onboarding dataset, class imbalance, cold-start distributions, and entity graph linkages.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df_m = pd.read_csv('../data/synthetic/merchants.csv')
df_r = pd.read_csv('../data/synthetic/relationships.csv')
print(f"Total Merchants: {len(df_m)}")
print(f"Fraud Rate: {df_m['fraud_label'].mean():.2%}")
print(f"Cold Start Merchants (tx count = 0): {(df_m['transaction_count'] == 0).mean():.2%}")
df_m.head()
```

### 1. Business Category Distribution & Class Balance

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=df_m, x='business_category', hue='fraud_label', palette='Set2')
plt.title('Merchant Distribution by Category and Fraud Label')
plt.show()
```

### 2. Cold-Start vs Established Merchants

```python
plt.figure(figsize=(8, 4))
df_m['is_cold_start'] = df_m['transaction_count'] == 0
sns.barplot(data=df_m, x='is_cold_start', y='fraud_label', estimator=np.mean)
plt.title('Fraud Probability: Cold-Start vs Established')
plt.ylabel('Fraud Rate')
plt.show()
```

### 3. Entity Linkage Cardinality & Multi-Tenant Sharing

```python
print('Entity Link Types:')
print(df_r['entity_type'].value_counts())

bank_sharing = df_m['bank_account_id'].value_counts()
print(f'Max merchants sharing a single bank: {bank_sharing.max()}')
print(f'Banks shared across >1 merchant: {(bank_sharing > 1).sum()}')
```
